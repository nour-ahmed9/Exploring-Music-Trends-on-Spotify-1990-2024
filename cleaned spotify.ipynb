```python
# Drop the 'Text After Delimiter' column
sp = sp.drop('Text After Delimiter', axis=1)
# Display the columns to confirm the column has been dropped
display(sp.columns)
```

```python
display(sp['Genre'].value_counts())
```

```python
# Get the unique values as a NumPy array
unique_genres_array = sp['Genre'].unique()

# Convert the NumPy array to a Python list
unique_genres_list = unique_genres_array.tolist()

# Print the list of unique genres
print(unique_genres_list)
```

```python
# Get the value counts of the 'Genre' column
genre_counts = sp['Genre'].value_counts()

# Filter for genres with a count of 3 or less
low_frequency_genres = genre_counts[genre_counts <= 3]

# Display the low frequency genres
display(low_frequency_genres)
```
